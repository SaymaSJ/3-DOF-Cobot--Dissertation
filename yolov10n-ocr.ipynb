```python
import re

def extract_date_candidates(text):
    patterns = [
        r'\b\d{4}[./-]\d{1,2}[./-]\d{1,2}\b',   # YYYY.MM.DD
        r'\b\d{1,2}[./-]\d{1,2}[./-]\d{2,4}\b', # DD.MM.YY or DD/MM/YYYY
        r'\b\d{1,2}[./-]\d{4}\b',               # MM/YYYY
        r'\b\d{8}\b'                             # YYYYMMDD or DDMMYYYY
    ]

    matches = []

    for pattern in patterns:
        matches.extend(re.findall(pattern, text))

    return matches
```

```python
import pandas as pd
import re

GT_FILE = "/content/ocr_verification_sorted_v2.csv"

df = pd.read_csv(GT_FILE).fillna("")

def normalize_date(text):
    text = str(text).strip().upper()

    if text in ["", "UNREADABLE", "NO_EXPLICIT_DATE"]:
        return ""

    # normalize separators
    text = text.replace("-", "/").replace(".", "/")

    # remove extra spaces
    text = re.sub(r"\s+", "", text)

    return text

df["gt_norm"] = df["ground_truth_date"].apply(normalize_date)

# extract likely date from OCR text
def extract_ocr_date(text):
    text = str(text)

    patterns = [
        r'\b\d{4}[./-]\d{1,2}[./-]\d{1,2}\b',
        r'\b\d{1,2}[./-]\d{1,2}[./-]\d{2,4}\b',
        r'\b\d{1,2}[./-]\d{4}\b'
    ]

    for p in patterns:
        m = re.search(p, text)
        if m:
            return normalize_date(m.group())

    return ""

df["ocr_norm"] = df["ocr_text"].apply(extract_ocr_date)

# only evaluate rows where you entered a readable ground-truth date
valid = df[df["gt_norm"] != ""].copy()

valid["exact_match"] = valid["ocr_norm"] == valid["gt_norm"]

correct = int(valid["exact_match"].sum())
total = len(valid)

print("Readable ground-truth images:", total)
print("Correct exact dates:", correct)

if total > 0:
    print(f"OCR exact-date accuracy: {100*correct/total:.1f}%")
```

```python
import re
import pandas as pd
from datetime import datetime

MONTHS = {
    "jan": 1, "january": 1,
    "feb": 2, "february": 2,
    "mar": 3, "march": 3,
    "apr": 4, "april": 4,
    "may": 5,
    "jun": 6, "june": 6,
    "jul": 7, "july": 7,
    "aug": 8, "august": 8,
    "sep": 9, "sept": 9, "september": 9,
    "oct": 10, "october": 10,
    "nov": 11, "november": 11,
    "dec": 12, "december": 12
}

def parse_ground_truth(gt):
    """
    Convert ground truth into a canonical representation.

    Full date -> ('FULL', year, month, day)
    Month/year -> ('MONTH', year, month)
    """
    gt = str(gt).strip().lower()

    # DD/MM/YYYY, DD-MM-YYYY, DD.MM.YYYY
    m = re.fullmatch(
        r'(\d{1,2})[./-](\d{1,2})[./-](\d{2,4})',
        gt
    )

    if m:
        day, month, year = map(int, m.groups())

        if year < 100:
            year += 2000

        return ("FULL", year, month, day)

    # DD-Mon-YY/YYYY
    m = re.fullmatch(
        r'(\d{1,2})[- /.]([a-z]+)[- /.](\d{2,4})',
        gt
    )

    if m:
        day = int(m.group(1))
        month_name = m.group(2)
        year = int(m.group(3))

        if month_name in MONTHS:
            if year < 100:
                year += 2000

            return ("FULL", year, MONTHS[month_name], day)

    # Mon-YY / Mon-YYYY
    m = re.fullmatch(
        r'([a-z]+)[- /.](\d{2,4})',
        gt
    )

    if m:
        month_name = m.group(1)
        year = int(m.group(2))

        if month_name in MONTHS:
            if year < 100:
                year += 2000

            return ("MONTH", year, MONTHS[month_name])

    # MM/YYYY
    m = re.fullmatch(r'(\d{1,2})[./-](\d{4})', gt)

    if m:
        month, year = map(int, m.groups())
        return ("MONTH", year, month)

    return None
```

```python
def ocr_candidates(text):

    text = str(text).lower()
    candidates = []

    # DD/MM/YYYY etc.
    for d, m, y in re.findall(
        r'\b(\d{1,2})[./-](\d{1,2})[./-](\d{2,4})\b',
        text
    ):
        d, m, y = int(d), int(m), int(y)

        if y < 100:
            y += 2000

        if 1 <= d <= 31 and 1 <= m <= 12:
            candidates.append(("FULL", y, m, d))

    # MM/YYYY
    for m, y in re.findall(
        r'\b(\d{1,2})[./-](20\d{2})\b',
        text
    ):
        m, y = int(m), int(y)

        if 1 <= m <= 12:
            candidates.append(("MONTH", y, m))

    # Month-name formats: AUG 23, Feb 2024 etc.
    for month_name, year in re.findall(
        r'\b(' + "|".join(MONTHS.keys()) +
        r')[\s./-]*(\d{2,4})\b',
        text
    ):
        y = int(year)

        if y < 100:
            y += 2000

        candidates.append(
            ("MONTH", y, MONTHS[month_name])
        )

    # DD Month YYYY
    for day, month_name, year in re.findall(
        r'\b(\d{1,2})[\s./-]+(' +
        "|".join(MONTHS.keys()) +
        r')[\s./-]+(\d{2,4})\b',
        text
    ):
        d = int(day)
        y = int(year)

        if y < 100:
            y += 2000

        candidates.append(
            ("FULL", y, MONTHS[month_name], d)
        )

    return candidates
```

```python
def is_exact_match(gt, ocr_text):

    gt_parsed = parse_ground_truth(gt)

    if gt_parsed is None:
        return False

    candidates = ocr_candidates(ocr_text)

    return gt_parsed in candidates


valid = df[
    ~df["ground_truth_date"]
      .astype(str)
      .str.upper()
      .isin(["", "UNREADABLE", "NO_EXPLICIT_DATE"])
].copy()

valid["exact_match_v2"] = valid.apply(
    lambda r: is_exact_match(
        r["ground_truth_date"],
        r["ocr_text"]
    ),
    axis=1
)

correct = int(valid["exact_match_v2"].sum())
total = len(valid)

print("Readable ground-truth images:", total)
print("Correct exact expiry dates:", correct)
print(f"OCR exact-date accuracy: {100*correct/total:.1f}%")
```

```python
import re
import calendar
import pandas as pd

MONTHS = {
    "jan":1, "january":1,
    "feb":2, "february":2,
    "mar":3, "march":3,
    "apr":4, "april":4,
    "may":5,
    "jun":6, "june":6,
    "jul":7, "july":7,
    "aug":8, "august":8,
    "sep":9, "sept":9, "september":9,
    "oct":10, "october":10,
    "nov":11, "november":11,
    "dec":12, "december":12
}

MONTH_PATTERN = "|".join(
    sorted(MONTHS.keys(), key=len, reverse=True)
)

def valid_full(y, m, d):
    try:
        y, m, d = int(y), int(m), int(d)

        if y < 100:
            y += 2000

        calendar.datetime.date(y, m, d)
        return ("FULL", y, m, d)

    except:
        return None
```

```python
from datetime import date

def make_full(y, m, d):
    try:
        y = int(y)
        m = int(m)
        d = int(d)

        if y < 100:
            y += 2000

        date(y, m, d)

        return ("FULL", y, m, d)

    except:
        return None
```

```python
def parse_ground_truth(text):

    s = str(text).strip().lower()

    # YYYY.MM.DD / YYYY-MM-DD / YYYY/MM/DD
    m = re.fullmatch(
        r'(\d{4})[./-](\d{1,2})[./-](\d{1,2})',
        s
    )
    if m:
        return make_full(m.group(1), m.group(2), m.group(3))

    # DD.MM.YYYY etc.
    m = re.fullmatch(
        r'(\d{1,2})[./-](\d{1,2})[./-](\d{2,4})',
        s
    )
    if m:
        return make_full(m.group(3), m.group(2), m.group(1))

    # DD-Mon-YY
    m = re.fullmatch(
        rf'(\d{{1,2}})[ ./-]*({MONTH_PATTERN})[ ./-]*(\d{{2,4}})',
        s
    )
    if m:
        y = int(m.group(3))
        if y < 100:
            y += 2000

        return make_full(
            y,
            MONTHS[m.group(2)],
            int(m.group(1))
        )

    # Mon-YY: Dec-23
    m = re.fullmatch(
        rf'({MONTH_PATTERN})[ ./-]*(\d{{2,4}})',
        s
    )
    if m:
        y = int(m.group(2))
        if y < 100:
            y += 2000

        return ("MONTH", y, MONTHS[m.group(1)])

    # YY-Mon: 23-Dec
    m = re.fullmatch(
        rf'(\d{{2,4}})[ ./-]*({MONTH_PATTERN})',
        s
    )
    if m:
        y = int(m.group(1))
        if y < 100:
            y += 2000

        return ("MONTH", y, MONTHS[m.group(2)])

    # MM/YYYY
    m = re.fullmatch(
        r'(\d{1,2})[./-](\d{4})',
        s
    )
    if m:
        month = int(m.group(1))
        year = int(m.group(2))

        if 1 <= month <= 12:
            return ("MONTH", year, month)

    return None
```

```python
def ocr_candidates(text):

    s = str(text).lower()
    candidates = []

    # YYYY MM DD
    # allows ., /, -, or spaces
    for y, m, d in re.findall(
        r'(?<!\d)(20\d{2})[ ./-]+(\d{1,2})[ ./-]+(\d{1,2})',
        s
    ):
        c = make_full(y, m, d)
        if c:
            candidates.append(c)

    # DD MM YYYY
    for d, m, y in re.findall(
        r'(?<!\d)(\d{1,2})[./-]+(\d{1,2})[./-]+(20\d{2})(?!\d)',
        s
    ):
        c = make_full(y, m, d)
        if c:
            candidates.append(c)

    # DD Month YYYY
    # also catches 16JAN2024
    for d, mon, y in re.findall(
        rf'(?<!\d)(\d{{1,2}})\s*({MONTH_PATTERN})\s*(20\d{{2}})',
        s
    ):
        c = make_full(y, MONTHS[mon], d)
        if c:
            candidates.append(c)

    # Month YYYY
    for mon, y in re.findall(
        rf'({MONTH_PATTERN})[\s./-]*(20\d{{2}})',
        s
    ):
        candidates.append(
            ("MONTH", int(y), MONTHS[mon])
        )

    # MM/YYYY
    for m, y in re.findall(
        r'(?<!\d)(\d{1,2})[./-](20\d{2})(?!\d)',
        s
    ):
        m = int(m)

        if 1 <= m <= 12:
            candidates.append(
                ("MONTH", int(y), m)
            )

    return list(dict.fromkeys(candidates))
```

```python
valid = df[
    ~df["ground_truth_date"]
      .astype(str)
      .str.strip()
      .str.upper()
      .isin(["", "UNREADABLE", "NO_EXPLICIT_DATE"])
].copy()

valid["gt_parsed"] = valid["ground_truth_date"].apply(
    parse_ground_truth
)

valid["ocr_candidates"] = valid["ocr_text"].apply(
    ocr_candidates
)

valid["exact_match_v3"] = valid.apply(
    lambda r:
        r["gt_parsed"] is not None
        and r["gt_parsed"] in r["ocr_candidates"],
    axis=1
)

correct = int(valid["exact_match_v3"].sum())
total = len(valid)

print("Readable ground-truth images:", total)
print("Ground truths parser could understand:",
      valid["gt_parsed"].notna().sum())
print("Correct exact expiry dates:", correct)
print(f"OCR exact-date accuracy: {100*correct/total:.1f}%")
```

```python
import cv2
import easyocr
import numpy as np
from PIL import Image
import os

reader = easyocr.Reader(['en'], gpu=True)

def improved_ocr(image_path):
    img = cv2.imread(image_path)

    # 1. Upscale substantially - expiry printing is often tiny
    img = cv2.resize(
        img, None,
        fx=3, fy=3,
        interpolation=cv2.INTER_CUBIC
    )

    # 2. Grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 3. Improve local contrast
    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )
    gray = clahe.apply(gray)

    # 4. Slight denoising
    gray = cv2.GaussianBlur(gray, (3, 3), 0)

    # 5. OCR tuned for dates
    result = reader.readtext(
        gray,
        detail=0,
        paragraph=False,
        allowlist=
        '0123456789'
        'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
        'abcdefghijklmnopqrstuvwxyz'
        './-: ',
        mag_ratio=2.0,
        contrast_ths=0.05,
        adjust_contrast=0.7,
        text_threshold=0.4,
        low_text=0.3
    )

    return " ".join(result)
```

```python
import re
from datetime import date

MONTHS = {
    "JAN": 1, "JANUARY": 1,
    "FEB": 2, "FEBRUARY": 2,
    "MAR": 3, "MARCH": 3,
    "APR": 4, "APRIL": 4,
    "MAY": 5,
    "JUN": 6, "JUNE": 6,
    "JUL": 7, "JULY": 7,
    "AUG": 8, "AUGUST": 8,
    "SEP": 9, "SEPT": 9, "SEPTEMBER": 9,
    "OCT": 10, "OCTOBER": 10,
    "NOV": 11, "NOVEMBER": 11,
    "DEC": 12, "DECEMBER": 12
}

MONTH_PATTERN = "|".join(
    sorted(MONTHS.keys(), key=len, reverse=True)
)

def make_full(y, m, d):
    try:
        y = int(y)
        m = int(m)
        d = int(d)

        if y < 100:
            y += 2000

        date(y, m, d)
        return ("FULL", y, m, d)

    except:
        return None


def ocr_candidates(text):

    s = str(text).upper()
    candidates = []

    # YYYY.MM.DD / YYYY-MM-DD / YYYY MM DD
    for y, m, d in re.findall(
        r'(?<!\d)(20\d{2})[\s./-]+(\d{1,2})[\s./-]+(\d{1,2})(?!\d)',
        s
    ):
        c = make_full(y, m, d)
        if c:
            candidates.append(c)

    # DD.MM.YYYY / DD-MM-YYYY / DD/MM/YYYY
    for d, m, y in re.findall(
        r'(?<!\d)(\d{1,2})[\s./-]+(\d{1,2})[\s./-]+(20\d{2})(?!\d)',
        s
    ):
        c = make_full(y, m, d)
        if c:
            candidates.append(c)

    # 25JUN2023 / 25 JUN 2023
    for d, mon, y in re.findall(
        rf'(?<!\d)(\d{{1,2}})\s*({MONTH_PATTERN})\s*(\d{{2,4}})',
        s
    ):
        year = int(y)
        if year < 100:
            year += 2000

        c = make_full(year, MONTHS[mon], d)

        if c:
            candidates.append(c)

    # AUG23 / AUG 23 / AUG2023
    for mon, y in re.findall(
        rf'({MONTH_PATTERN})[\s./-]*(\d{{2,4}})(?!\d)',
        s
    ):
        year = int(y)

        if year < 100:
            year += 2000

        if 2000 <= year <= 2099:
            candidates.append(
                ("MONTH", year, MONTHS[mon])
            )

    # MM/YYYY
    for m, y in re.findall(
        r'(?<!\d)(\d{1,2})[./-](20\d{2})(?!\d)',
        s
    ):
        month = int(m)

        if 1 <= month <= 12:
            candidates.append(
                ("MONTH", int(y), month)
            )

    return list(dict.fromkeys(candidates))
```
